```python
# 3

def merge(a: list, b: list) -> list:
    return sorted(a + b)

print(merge([1, 2, 3], [-1, 2, 4]))
```

```python
# 4

def count_even_odd(lst: list):
    even = sum(1 - elem % 2 for elem in lst)
    odd = sum(elem % 2 for elem in lst)
    print('even:', even)
    print('odd:', odd)
    
count_even_odd([1, 1, 4, 5, 1, 4])
```

```python
# 5

def is_valid(inpt: str) -> bool:
    if len(inpt) < 6 or len(inpt) > 16:
        return False
    if sum(ch in '$#@' for ch in inpt) == 0:
        return False
    if sum('0' <= ch <= '9' for ch in inpt) == 0:
        return False
    if sum('a' <= ch <= 'z' for ch in inpt) == 0:
        return False
    if sum('A' <= ch <= 'Z' for ch in inpt) == 0:
        return False
    return True

print(is_valid('IloveCs1602!#'))
print(is_valid('Hello World9'))
```

```python
# 6

def vowel_or_consonant(ch: str) -> str:
    assert len(ch) == 1
    ch = ch.lower()
    if ch in 'aeiou':
        return 'vowel'
    return 'consonant'

print(vowel_or_consonant('e'))
print(vowel_or_consonant('C'))
print(vowel_or_consonant('A'))
```

```python
# 7

def month2days(mon: str) -> int | str:
    days = {
        'January': 31,
        'February': '28 or 29',
        'March': 31,
        'April': 30,
        'May': 31,
        'June': 30,
        'July': 31,
        'August': 31,
        'September': 30,
        'October': 31,
        'November': 30,
        'December': 31,
    }
    if days.get(mon):
        return days[mon]
    return 'invaid month name'

print(month2days('Eember'))
print(month2days('February'))
print(month2days('December'))
```

```python
# 8

def check_int(s: str) -> bool:
    try:
        int(s)
    except:
        return False
    return True
    
print(check_int('114514'))
print(check_int('1919.810'))
```

```python
# 9

def get_season(mon: int, day: int):
    date = str(mon).zfill(2) + str(day).zfill(2)
    if date < '0204':
        print('Winter')
    elif date < '0505':
        print('Spring')
    elif date < '0807':
        print('Summer')
    elif date < '1107':
        print('Autumn')
    else:
        print('Winter')

get_season(8, 7)
```

```python
# 11

def is_leap_year(yr):
    return (yr % 4 == 0 and (yr % 100 != 0 or yr % 400 == 0))

def get_next_day(yr, mon, day):
    days = [-1, 31, 29 if is_leap_year(yr) else 28, 31, 30, 31, 30, 31, 31, 30, 31, 30, 31]
    day += 1
    if day > days[mon]:
        mon += 1
        day = 1
        if mon > 12:
            mon = 1
            yr += 1
    return f'The next date is [yyyy-mm-dd]: {yr}-{mon}-{day}'
    
yr = int(input('Input a year: '))
mon = int(input('Input a month [1-12]: '))
day = int(input('Input a day [1-31]: '))
print(get_next_day(yr, mon, day))
```
