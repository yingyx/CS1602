```python
# 6
import math

def area(a, b, c):
    p = (a + b + c) / 2
    return math.sqrt(p * (p - a) * (p - b) * (p - c))

def angle(a, b, c):
    ca = (c * c + b * b - a * a)/(2 * c * b)
    cb = (a * a + c * c - b * b)/(2 * a * c)
    cc = (a * a + b * b - c * c)/(2 * b * a)
    return math.acos(ca), math.acos(cb), math.acos(cc)

def circumcircle(a, b, c):
    r1 = a / (2 * math.sin(math.acos((c * c + b * b - a * a)/(2 * c * b))))
    return r1, math.pi * r1 * r1

def incircle(a, b, c):
    p = (a + b + c) / 2
    r2 = area(a, b, c) / p
    return r2, math.pi * r2 * r2

print(area(3, 4, 6))
print(angle(3, 4, 6))
print(circumcircle(3, 4, 6))
print(incircle(3, 4, 6))
```

```python
# 7
import math

def intersections(x1, y1, r1, x2, y2, r2):
    dis = math.sqrt((x1 - x2) ** 2 + (y1 - y2) ** 2)
    if r1 + r2 < dis or abs(r1 - r2) > dis:
        return 0
    if abs(r1 - r2) == dis or r1 + r2 == dis:
        return 1
    return 2
print(intersections(-1, 1, 1, 1, 1, 1))
```
