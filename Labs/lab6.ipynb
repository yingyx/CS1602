```python
# 1

def f(n):
    if n == 0 or n == 1:
        return 1
    return f(f(n // 2)) + 1

print(f(5))
```

```python
# 2

from functools import cache

# This is unbelievably slow!

@cache
def a1(n):
    if n == 0 or n == 1:
        return 1
    return a1(n - 1) + a1(n - 2)

_a1 = [1, 1]
for i in range(2, 1001):
    _a1.append(_a1[i - 1] + _a1[i - 2])
    
print(a1(1000), _a1[1000])

@cache
def a2(n):
    if n == 0 or n == 1 or n == 2:
        return 1
    return a2(n - 1) + 2 * a2(n - 2) - a2(n - 3)

_a2 = [1, 1, 1]
for i in range(3, 1001):
    _a2.append(_a2[i - 1] + 2 * _a2[i - 2] - _a2[i - 3])
    
print(a2(1000), _a2[1000])

@cache
def a3(n):
    if n == 0:
        return 2
    if n == 1:
        return 3
    return a3(n - 1) * a3(n - 2)

_a3 = [2, 3]
for i in range(2, 1001):
    _a3.append(_a3[i - 1] * _a3[i - 2])
    
print(a3(1000), _a3[1000])

@cache
def a4(m, n):
    if m == 0 or n == 0:
        return 1
    return a4(m, n - 1) + a4(m - 1, n) - a4(m - 1, n - 1)

_a4 = [[[] for _ in range(101)] for __ in range(101)]
for i in range(101):
    for j in range(101):
        if i == 0 or j == 0:
            _a4[i][j] = 1
        else:
            _a4[i][j] = _a4[i - 1][j] + _a4[i][j - 1] - _a4[i - 1][j - 1]
            
print(a4(100, 100), _a4[100][100])
```

```python
# 3

def get_sum(lst):
    sum = 0
    for elem in lst:
        if type(elem) == list:
            sum += get_sum(elem)
        else:
            sum += elem
    return sum

print(get_sum([1, 2, [3, 4, [5, 6]]]))    
```

```python
# 4

def combine_dic(dic1, dic2):
    new_dic = {}
    for key in dic1:
        if new_dic.get(key):
            new_dic[key] += dic1[key]
        else:
            new_dic[key] = dic1[key]
    for key in dic2:
        if new_dic.get(key):
            new_dic[key] += dic2[key]
        else:
            new_dic[key] = dic2[key]
    return new_dic

def get_count(lst):
    dic = {}
    for elem in lst:
        if type(elem) == list:
            dic = combine_dic(dic, get_count(elem))
        else:
            if dic.get(elem):
                dic[elem] += 1
            else:
                dic[elem] = 1
    return dic

print(get_count([1, 2, 3, 4, 5, 6, [3, 4, 5, 6, [5, 6]]]))
```

```python
# 5

import math

class Hadamard:
    def __init__(self, k = -1, mat = [[1]]):
        if k == 0 or mat != [[1]]:
            self.mat: list[list] = mat
        else:
            self.mat = (Hadamard(k - 1) + Hadamard(k - 1)) + (Hadamard(k - 1) + Hadamard(k - 1).neg())
        self.width: int = len(self.mat)
        self.height: int = self.width
        self.k = k
        if k == -1:
            self.k = int(math.log2(self.width))
        
    def __add__(self, x):
        if self.width == x.width:
            new_a = []
            for i in range(self.width):
                new_a.append(self.mat[i] + x.mat[i])
            return new_a
        else:
            new_a = []
            for i in range(self.height + x.height):
                if i < self.height:
                    new_a.append(self.mat[i])
                else:
                    new_a.append(x.mat[i - self.height])
            return new_a    
    
    def __str__(self):
        output = ''
        for i in range(self.width):
            for j in range(self.height):
                if self.mat[i][j] == 1:
                    output += ' '
                output += str(self.mat[i][j]) + ' '
            output += '\n'
        return output

    def neg(self):
        new = []
        for i in range(self.width):
            row = []
            for j in range(self.width):
                row.append(-self.mat[i][j])
            new.append(row)
        return Hadamard(self.k, new)
    
print(Hadamard(3))
```

```python
# 6

def Powerset(s: set):
    s = list(s)
    if not s:
        return [[]]
    first, other = s[0], s[1:]
    s1 = Powerset(other)
    s2 = [[first] + ss for ss in s1]
    return s1 + s2
        
print(Powerset(set([1, 6, 8, 10])))
```

```python
# 7

def solve(row, cols: set, diag1: set, diag2: set, map: list[list]) -> None:
    if row == 8:
        solutions.append(["".join(row) for row in map])
        return
    
    for col in range(8):
        
        if col in cols or (row - col) in diag1 or (row + col) in diag2:
            continue
        
        map[row][col] = 'Q'
        cols.add(col)
        diag1.add(row - col)
        diag2.add(row + col)
        
        solve(row + 1, cols, diag1, diag2, map)
        
        map[row][col] = '.'
        cols.remove(col)
        diag1.remove(row - col)
        diag2.remove(row + col)
    
solutions = []
solve(0, set(), set(), set(), [["." for __ in range(8)] for _ in range(8)])

for solution in solutions:
    print("\n".join(solution))
    print("=" * 8)
```
