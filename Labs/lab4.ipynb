```python
# 1

def sum_and_prod(lst: list[int]):
    sum, prod = 0, 1
    for elem in lst:
        sum += elem
        prod *= elem
    return sum, prod

print(sum_and_prod([1, 1, 4, 5, 1, 100]))
```

```python
# 2

def compare_list(lst1: list, lst2: list) -> tuple[bool, list]:
    if all(elem in lst1 for elem in lst2):
        return True, lst1
    if all(elem in lst2 for elem in lst1):
        return True, lst2
    return False, []
    
def compare_list2(lst1: list, lst2: list) -> tuple[bool, list]: # Alternative using set union operand
    if set(lst1) | set(lst2) == set(lst1):
        return True, lst1
    if set(lst1) | set(lst2) == set(lst2):
        return True, lst2
    return False, []

print(compare_list([2, 3, 4], [1, 2, 3]))
print(compare_list2([2, 3, 4], [1, 2, 3]))
print(compare_list([1, 1, 4], [5, 1, 4]))
print(compare_list2([1, 1, 4], [5, 1, 4]))
```

```python
# 3

def is_all_even(num: int) -> bool:
    while num:
        if (num % 10) % 2 == 1:
            return False
        num //= 10
    return True

for i in range(1000, 3001):
    if is_all_even(i):
        print(i)
```

```python
# 4

def remove_duplicate(lst: list) -> list:
    new_lst = []
    for elem in lst:
        if not elem in new_lst:
            new_lst.append(elem)
    return new_lst

def remove_duplicate2(lst: list) -> list:
    return list(set(lst))

print(remove_duplicate([12, 24, 35, 24, 88, 120, 155, 88, 120, 155]))
print(remove_duplicate2([12, 24, 35, 24, 88, 120, 155, 88, 120, 155]))
```

```python
# 5

def flip_number(num: int) -> int:
    k = -1 if num < 0 else 1
    num = str(abs(num))
    new_num = []
    for i in range(0, len(num)):
        new_num.append(num[-i-1])
    return k * int(''.join(new_num))

print(flip_number(-114))
print(flip_number(514))    
```

```python
# 6

def find_intersection(lst1: list, lst2: list):
    inter = []
    for elem in lst1:
        if elem in lst2:
            inter.append(elem)
    return inter

def find_intersection2(lst1: list, lst2: list):
    inter = list(set(lst1) & set(lst2))
    if inter:
        return inter
    else:
        print("no intersection")
        return False
    
print(find_intersection([123, 456], [345, 456]))
print(find_intersection2([123, 456], [345, 456]))
    
```

```python
# 7

def print_star(n: int):
    for i in range(2 * n):
        print('*' * ((i + 1)if i < n else 2 * n - i - 1))
        
print_star(5)
print_star(6)
```

```python
# 8

def count_digits(n):
    lst = [0] * 10
    for ch in str(n):
        lst[ord(ch) - ord('0')] += 1
    return lst

print(count_digits(12333219))
print(count_digits(1145141919810))
```

```python
# 9

def sum_of_two(n: int):
    return sum(2 * (n - i) * (10 ** i) for i in range(n))

print(sum_of_two(4))
print(sum_of_two(114))
print(sum_of_two(514))
```

```python
# 11

def find_min(lst: list) -> int:
    r, l = 0, len(lst) - 1
    for i in range(len(lst) - 1):
        if lst[i] > lst[i + 1]:
            r = i - 1
            break
    for i in range(len(lst) - 1, 0, -1):
        if lst[i - 1] > lst[i]:
            l = i
            break
    return min(l, r)


print(find_min([1, 2, 100, 800, 800, 514, 200, 114, 99, 1]))
```

```python
# 12

import math

def is_prime(n: int) -> bool:
    for i in range(2, math.floor(n ** 0.5 + 1)):
        if n % i == 0:
            return False
    return True

def sieve_of_eratosthenes():
    prime = [True] * 1000000
    prime[0], prime[1] = False, False
    for i in range(2, 1000000):
        if not prime[i]:
            continue
        prime[i] = is_prime(i)
        for j in range(i * 2, 1000000, i):
            prime[j] = False
    for i, isp in enumerate(prime):
        if isp:
            print(i, end=' ')
            
# sieve_of_eratosthenes()
```
