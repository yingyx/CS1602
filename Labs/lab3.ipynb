```python
# 1

def is_triangle(a, b, c) -> bool:
    return a + b > c and a + c > b and b + c > a

print(is_triangle(1, 4, 5))
print(is_triangle(2, 4, 5))
```

```python
# 2

def delta(a, b, c) -> tuple[int, list]:
    if a == 0:
        if b == 0:
            if c == 0:
                return 'inf', []
            return 0, []
        return 1, [-c / b]
    # a != 0
    delt = b * b - 4 * a * c
    if delt > 0:
        return 2, [(-b + (delt ** 0.5)) / (2*a), (-b - (delt ** 0.5)) / (2*a)]
    if delt == 0:
        return 1, [-b / (2*a)]
    return 0, []

print(delta(0, 0, 0))
print(delta(0, 0, 1))
print(delta(0, 1, 1))
print(delta(1, 2, 1))
print(delta(1, 7, 4))
```

```python
# 3

import math

def is_prime(n) -> bool:
    for i in range(2, math.floor(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True

print(is_prime(2))
print(is_prime(3))
print(is_prime(25))
print(is_prime(47))
```

```python
# 4

def is_palindrome_year(year):
    year = str(year)
    for i in range(len(year) // 2):
        if year[i] != year[len(year) - 1 - i]:
            return False
    return True
    # Alternative:
    # return year[::-1] == year_str

print(is_palindrome_year(2444))
print(is_palindrome_year(999))
print(is_palindrome_year(2002))
```

```python
# 5

import math

def f(x):
    if x < -2:
        return x ** 4
    if x < 2:
        return math.sin(x)
    return math.exp(x)

print(f(-3))
print(f(0))
print(f(4))
```

```python
# 7

def count_end_zero(n: int) -> int:
    return len(str(n)) - len(str(n).rstrip('0'))

print(count_end_zero(1235712573192878000000))
print(count_end_zero(1000090))
```

```python
# 8

def hamming(a, b) -> int:
    return bin(a ^ b).count('1')

print(hamming(3, 7))
```

```python
# 11

def fractal(n): # Why not math.factorial?
    ans = 1
    while n > 0:
        ans *= n
        n -= 1
    return ans

def choose(n, k): # Why not math.comb?
    return fractal(n) // (fractal(k) * fractal(n- k))

def pascal(n):
    for i in range(n):
        print(' ' * (n - 1 - i), end='')
        for j in range(i + 1):
            print(choose(i, j), end=' ')
        print()
        
pascal(9)
```
