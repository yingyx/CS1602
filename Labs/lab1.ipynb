```python
# 3
def is_tri(a, b, c):
    return a + b > c and a + c > b and b + c > a and a > 0 and b > 0 and c > 0
def tri_area(a, b, c):
    if not is_tri(a, b, c):
        return
    q = (a + b + c) / 2
    return (q * (q - a) * (q - b) * (q - c)) ** 0.5
print('Area of triangle with edges: 3, 4, 3:', tri_area(3, 4, 3))
print('Area of triangle with edges: 1, 1, 1:', tri_area(1, 1, 1))
print('Area of triangle with edges: 2, 3, 2:', tri_area(2, 3, 2))
print('Area of triangle with edges: 3, 3, 1:', tri_area(3, 3, 1))
```

```python
# 4
def edge_length(x1, y1, x2, y2):
    return (((x1 - x2) ** 2) + ((y1 - y2) ** 2)) ** 0.5
def tri_area_by_coords(p1, p2, p3):
    x1, y1 = p1
    x2, y2 = p2
    x3, y3 = p3
    e1 = edge_length(x1, y1, x2, y2)
    e2 = edge_length(x2, y2, x3, y3)
    e3 = edge_length(x3, y3, x1, y1)
    return tri_area(e1, e2, e3)

print('Area of triangle with points: (0.1, 0.1), (0.3, 0.3), (0.4, 0.5):', tri_area_by_coords((.1, .1), (.3, .3), (.4, .5)))
```

```python
# 5
def calc_roots(a, b, c):
    print(f'a={a}, b={b}, c={c}: ', end='')
    if a == 0:
        if b == 0:
            if c:
                print('No solution.')
                return
            else:
                print('')
        else:
            print('x =', -c / b)
            return
    delta = b * b - 4 * a * c
    if delta < 0:
        print('No solution.')
    elif delta == 0:
        print('x1 = x2 =', -b / (2 * a))
    else:
        print('x1 =', (-b + delta ** 0.5) / (2 * a), ',', 'x2 =', (-b - delta ** 0.5) / (2 * a))

calc_roots(1, 2, -1)
calc_roots(1, -2, 1)
calc_roots(6, 6, 6)
calc_roots(0, 1, 1)
```

```python
# 6
def calc_maxin(a, b, c):
    if a == 0:
        return
    else:
        return (4 * a * c - b * b) / (4 * a)
    
print(calc_maxin(1, 1, 1))
print(calc_maxin(-1, 1, 1))
```
